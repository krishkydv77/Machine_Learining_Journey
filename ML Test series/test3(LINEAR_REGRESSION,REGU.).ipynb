```python
le=LabelEncoder()
for i in df.select_dtypes(include=['object']).columns:
  df[i]=le.fit_transform(df[i])
df.sample(5)
```

```python
x=df.drop(columns=['charges'])
y=df['charges']
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=23)
```

```python
ss=StandardScaler()
x_train=ss.fit_transform(x_train)
x_test=ss.transform(x_test)
```

```python
from sklearn.linear_model import LinearRegression, Ridge, Lasso, ElasticNet,LassoCV,ElasticNetCV,RidgeCV
from sklearn.metrics import accuracy_score, mean_absolute_error, mean_squared_error, r2_score
```

```python
#cv
ridge_cv = RidgeCV(alphas=[0.001, 0.01, 0.1, 1, 10], cv=5)
ridge_cv.fit(x_train, y_train)
print(" alpha  Ridge:", ridge_cv.alpha_)

lasso_cv = LassoCV(alphas=[0.001, 0.01, 0.1, 10], cv=5, random_state=42)
lasso_cv.fit(x_train, y_train)
print("alpha  Lasso:", lasso_cv.alpha_)

elastic_cv = ElasticNetCV(alphas=[0.001, 0.01, 0.1, 1, 10],
                          l1_ratio=[0.1, 0.3, 0.5, 0.9, 1],
                          cv=5,random_state=42)
elastic_cv.fit(x_train, y_train)

print("alpha  ElasticNet:", elastic_cv.alpha_)
print("l1_ratio  ElasticNet:", elastic_cv.l1_ratio_)
```

```python
models = {
    "Linear Regression": LinearRegression(),
    "Ridge Regression": Ridge(alpha=0.1),
    "Lasso Regression": Lasso(alpha=10.0),
    "ElasticNet Regression": ElasticNet(alpha=10.0, l1_ratio=0.1)
}

output = {}

for name, model in models.items():
    model.fit(x_train, y_train)
    y_pred = model.predict(x_test)
    
    output[name] = {
        "R2 Score": round(r2_score(y_test, y_pred), 2),
        "MSE": round(mean_squared_error(y_test, y_pred), 2),
        "MAE": round(mean_absolute_error(y_test, y_pred), 2),
        "RMSE": round(np.sqrt(mean_squared_error(y_test, y_pred)), 2)
    }
```

```python
df_new = pd.DataFrame(output).T
print(df_new)
```
