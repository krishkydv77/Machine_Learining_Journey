**Boosting:-**

- Boosting is an ensemble learning method
- which combines multiple weak or base learners to create a stronger predictive model.

- The main idea behind boosting is to sequentially train series of models

- where each subsequent model focuses on the instances that were misclassified by previous models.

- This iterative process allows the ensemble to learn from its mistakes and improve its overall performance.

**Boosting Process**

**Initialization →** each instance in the training set is assigned an equal weight.

**Iterative training →** A base learner is trained on the weighted training set and its predictions are evaluated.

**Weight update →** The weight of misclassified instances are increased, while correctly classified instances receive lower weights.

**Formation →** The trained base learner is added to the ensemble with a weight that depends on its performance.

**Termination →** Process continues until predefined iterations or performance threshold is reached.

**Final Prediction →** Ensemble combines weighted predictions of all base learners (weighted voting scheme).

**Types of Boosting**

**1.Adaptive Boosting (AdaBoost)**

Assigns higher weight to misclassified instances.

Sequentially trains weak learners and combines results.

Primarily used for binary classification problems.

**2.Gradient Boosting**

Builds ensemble of weak learners stage-wise.

Each subsequent model corrects mistakes of previous model by fitting negative gradient of loss function.

Can handle both classification and regression tasks.

**XGBoost →** Optimized GBM, parallel processing, regularization, missing values handling.

**LightGBM →** Faster training, lower memory usage, novel tree-growing algorithm.

**CatBoost →** Handles categorical features directly, ordered boosting, random permutation, symmetric trees.

| Feature | Bagging | Boosting |
| --- | --- | --- |
| Training | All models trained parallely | Models trained sequentially |
| Logic | Majority vote of independent models | Each model corrects mistakes of previous |
| Goal | Reduce variance | Reduce bias |
| Data Selection | Random sampling (Random Forest) | Misclassified data given higher weights (XGBoost, AdaBoost) |



**Model Selection**

Dataset size ~25,000 → **AdaBoost / Gradient Boosting**

Dataset size 5,000–100,000 → **XGBoost**

Dataset size >100,000 → **LightGBM**

Many categorical columns → **CatBoost**

**Avoid situations:-**

**AdaBoost →** noisy or very large datasets

**Gradient Boosting →** very large datasets (slow training)

**XGBoost →** too many categorical features (manual encoding needed)

**LightGBM →** very small datasets (overfitting risk)

**CatBoost →** purely numerical datasets

```python
from sklearn.ensemble import AdaBoostClassifier
from xgboost import XGBClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
```

```python
x,y=make_classification(n_samples=1000,n_features=10,random_state=42)
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
adaboost =AdaBoostClassifier(n_estimators=100,random_state=42)
adaboost.fit(x_train,y_train)

xgboost= XGBClassifier(n_estimators=100,random_state=42)
xgboost.fit(x_train,y_train)
```

```python
y_pred_adaboost=adaboost.predict(x_test)
y_pred_xgboost=xgboost.predict(x_test)
```

```python
accuracy_adaboost=accuracy_score(y_test,y_pred_adaboost)
accuracy_xgboost=accuracy_score(y_test,y_pred_xgboost)
```

```python
print("adaboost accuracy :",accuracy_adaboost)
print("xgboost accuracy :", accuracy_xgboost)
```
