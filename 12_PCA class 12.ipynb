# **PCA (Principal Component Analysis)**

## What is PCA?

**Answer:**

PCA (Principal Component Analysis) is an **unsupervised dimensionality reduction technique** that transforms a large set of correlated features into a smaller set of uncorrelated variables called **Principal Components**, while preserving as much information (variance) as possible.

---

## Why do we use PCA?

1. **Reduce dimensionality** of the dataset.
2. **Remove redundant features** (highly correlated variables).
3. **Reduce overfitting** in machine learning models.
4. **Improve model performance** and training speed.
5. **Help visualize high-dimensional data** in 2D or 3D.

---

# Working Process of PCA

## Step 1: Standardize the Data

Scale all features to a similar range so that each feature contributes equally.

* Mean = 0
* Standard Deviation = 1

This is usually done using StandardScaler.

---

## Step 2: Mean Center the Data

Calculate the mean of each feature and subtract it from every observation.

This shifts the data around the origin (0,0).

---

## Step 3: Compute the Covariance Matrix

The covariance matrix measures how features vary with respect to each other.

It helps us understand the relationships between features.

---

## Step 4: Calculate Eigenvalues and Eigenvectors

* **Eigenvectors** determine the directions of maximum variance.
* **Eigenvalues** determine how much variance exists in those directions.

---

## Step 5: Sort Eigenvalues in Descending Order

Arrange eigenvalues from highest to lowest.

Higher eigenvalues indicate more important components because they contain more information.

---

## Step 6: Form Principal Components

* Eigenvector with the **largest eigenvalue → PC1**
* Eigenvector with the **second largest eigenvalue → PC2**
* Continue similarly for the remaining components.

### Important:

**PC1 captures the maximum variance in the data.**
**PC2 captures the second-highest variance and is orthogonal (90°) to PC1.**

---

## Step 7: Select the Top Principal Components

Choose the components that preserve most of the variance.

Example:

| Component | Variance Explained |
| --------- | ------------------ |
| PC1       | 60%                |
| PC2       | 25%                |
| PC3       | 10%                |
| PC4       | 5%                 |

If PC1 + PC2 explain 85% variance, we can keep only these two components.

---

## Step 8: Transform the Original Data

Project the original data onto the selected principal components.

The result is a new dataset with fewer dimensions but most of the important information preserved.

---

# Interview Question

### What are Principal Components?

Principal Components are new features created by PCA that are linear combinations of the original features.

* They are uncorrelated with each other.
* PC1 captures maximum variance.
* PC2 captures the second maximum variance.
* Each subsequent component captures less variance.

---

```python
import numpy as np
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
from sklearn.ensemble import RandomForestClassifier
from sklearn.pipeline import Pipeline
from sklearn.preprocessing import StandardScaler
from sklearn.decomposition import PCA
```

```python
df=pd.read_csv('/content/Social_Network_Ads - Social_Network_Ads.csv')
df.head()
```

```python
df=df.drop(columns=['User ID','Gender'])
```

```python
x=df.drop(columns=['Purchased'],axis=1)
y=df['Purchased']
```

```python
from sklearn import pipeline
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
pipe=pipeline.Pipeline([('scaler',StandardScaler()),
                        ('pca',PCA(n_components=2)),
                        ('classifier',RandomForestClassifier())])
```

```python
pipe
```

```python
pipe.fit(x_train,y_train)
```

```python
y_pred=pipe.predict(x_test)
```

```python
print(accuracy_score(y_test,y_pred))
```
