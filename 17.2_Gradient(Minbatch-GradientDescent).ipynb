# Mini Batch Gradient Descent:- 
Mini Batch Gradient Descent is an optimization algorithm.

It updates the model weights using a small group (batch) of training samples, not the whole dataset and not just one sample.(25x4)

# Why do we use Mini Batch Gradient Descent?

We use Mini Batch GD because:

- Faster than Batch Gradient Descent
- More stable than SGD
- Uses less memory
- Trains large datasets efficiently
- Commonly used in Deep Learning


# When should we use Mini Batch GD?

Use Mini Batch GD when:

- Dataset is large

- Need faster training

- Want stable learning

- Training Neural Networks or Deep Learning models

# When should we avoid Mini Batch GD?

Avoid Mini Batch GD when:

- Dataset is very small (Batch GD is enough)

- Batch size is too large (becomes like Batch GD)

- Batch size is too small (becomes like SGD)

# Goal :-Minimize the loss function by updating weights using a small batch of samples.

# Advantages

- Faster than Batch GD

- More stable than SGD

- Uses less memory

- Good for large datasets

- Very popular in Deep Learning

- Better convergence

# Disadvantages

- Need to choose the correct batch size

- Larger batches need more memory

- Still requires learning rate tuning



| Feature       | Batch GD       | SGD                | Mini Batch GD                 |
| ------------- | -------------- | ------------------ | ----------------------------- |
| Data Used     | Entire dataset | One sample         | Small batch                   |
| Speed         | Slow           | Fast               | Fast                          |
| Memory        | High           | Low                | Medium                        |
| Weight Update | Once per epoch | After every sample | After every batch             |
| Stability     | Very Stable    | Noisy              | Stable                        |
| Best For      | Small dataset  | Very large dataset | Large dataset & Deep Learning |

```python
import random

class MBGDRegressor:
    
    def __init__(self,batch_size,learning_rate=0.01,epochs=100):
        
        self.coef_ = None
        self.intercept_ = None
        self.lr = learning_rate
        self.epochs = epochs
        self.batch_size = batch_size
        
    def fit(self,X_train,y_train):
        # init your coefs
        self.intercept_ = 0
        self.coef_ = np.ones(X_train.shape[1])
        
        for i in range(self.epochs):
            
            for j in range(int(X_train.shape[0]/self.batch_size)):
                
                idx = random.sample(range(X_train.shape[0]),self.batch_size)
                
                y_hat = np.dot(X_train[idx],self.coef_) + self.intercept_
                #print("Shape of y_hat",y_hat.shape)
                intercept_der = -2 * np.mean(y_train[idx] - y_hat)
                self.intercept_ = self.intercept_ - (self.lr * intercept_der)

                coef_der = -2 * np.dot((y_train[idx] - y_hat),X_train[idx])
                self.coef_ = self.coef_ - (self.lr * coef_der)
        
        print(self.intercept_,self.coef_)
    
    def predict(self,X_test):
        return np.dot(X_test,self.coef_) + self.intercept_
```
