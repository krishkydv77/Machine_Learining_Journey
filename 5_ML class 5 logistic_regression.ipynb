# How It Works?

Logistic Regression first calculates a value and then applies a Sigmoid Function.

The sigmoid function converts any value into a probability between 0 and 1.



# Logistic Regression Cheat Sheet

## What Is Logistic Regression?

Despite the name **"Regression"**, Logistic Regression is a **Classification Algorithm**.

It is mainly used to predict **binary outcomes (0 or 1)**.

**Examples:**

* Yes / No
* Spam / Not Spam
* Pass / Fail
* Fraud / Not Fraud

---

# How It Works 

## 1. Calculate the Linear Equation

The model first calculates a linear score (**z**).

### Formula

$[
\boxed{z = b_0 + b_1x_1 + b_2x_2 + \cdots + b_nx_n}
]$

Where:

* **b₀** = Intercept (Bias)
* **b₁, b₂, ...** = Weights (Coefficients)
* **x₁, x₂, ...** = Input features

**Purpose:**
Calculate a score (**z**) from the input features.

---

## 2. Apply the Sigmoid Function

The linear score (**z**) is converted into a probability between **0 and 1**.

### Formula

$[
\boxed{P=\frac{1}{1+e^{-z}}}
]$

Where:

* **P** = Predicted probability
* **e** = Euler's number (≈ 2.718)
* **z** = Linear equation output

**Purpose:**
Convert any value into a probability.

---

# Decision Rule (Threshold = 0.5)

* **Probability ≥ 0.5** → Class **1 (Yes)**
* **Probability < 0.5** → Class **0 (No)**

**Note:**
The threshold can be changed (for example, 0.3 or 0.7) depending on the problem.

---

# Loss Function

## Log Loss (Binary Cross-Entropy)

**What it does:**
Measures how different the predicted probabilities are from the actual values.

### Formula

$[
\boxed{
LogLoss=
-\frac{1}{n}
\sum
\left[
y\log(p)+(1-y)\log(1-p)
\right]
}
]$

Where:

* **y** = Actual class (0 or 1)
* **p** = Predicted probability
* **n** = Number of observations

**Goal:**
Minimize Log Loss to improve the model.

---

# Key Performance Metrics

## 1. Confusion Matrix

**What it does:**
Compares actual values with predicted values.

| Actual / Predicted | Positive | Negative |
| ------------------ | -------- | -------- |
| **Positive**       | TP       | FN       |
| **Negative**       | FP       | TN       |

Where:

* **TP** = True Positive
* **TN** = True Negative
* **FP** = False Positive
* **FN** = False Negative

---

## 2. Accuracy

**What it does:**
Measures the percentage of correct predictions.

### Formula

$[
\boxed{
Accuracy=
\frac{TP+TN}
{TP+TN+FP+FN}
}
]$

**Best Use:**
Balanced datasets.

---

## 3. Precision

**What it does:**
Out of all predicted positives, how many are actually positive?

### Formula

$[
\boxed{
Precision=
\frac{TP}
{TP+FP}
}
]$

**Best Use:**
When **False Positives** are costly.

---

## 4. Recall (Sensitivity)

**What it does:**
Out of all actual positives, how many did the model correctly identify?

### Formula

$[
\boxed{
Recall=
\frac{TP}
{TP+FN}
}
]$

**Best Use:**
When **False Negatives** are costly.

---

## 5. F1-Score

**What it does:**
Balances Precision and Recall into a single score.

### Formula
$
[
\boxed{
F1=
2\times
\frac{Precision\times Recall}
{Precision+Recall}
}
]$

**Best Use:**
Imbalanced datasets.

---

# Assumptions of Logistic Regression

### 1. Binary Target

The target variable should have **two classes** (0/1).

---

### 2. Independent Observations

Each observation should be independent of the others.

---

### 3. No Multicollinearity

Input features should not be highly correlated.

**Check using:** VIF (Variance Inflation Factor)

---

### 4. Linear Relationship with Log Odds

The input features should have a linear relationship with the **log odds** of the target.

---

# Quick Revision

| Topic                | Formula                                                  | Purpose                      |
| -------------------- | -------------------------------------------------------- | ---------------------------- |
| **Linear Equation**  | $(z=b_0+b_1x_1+\cdots+b_nx_n) $                            | Calculate score              |
| **Sigmoid Function** | $(\frac{1}{1+e^{-z}})$                                     | Convert score to probability |
| **Accuracy**         | $(\frac{TP+TN}{TP+TN+FP+FN})$                              | Overall correct predictions  |
| **Precision**        | $(\frac{TP}{TP+FP})$                                       | Correct positive predictions |
| **Recall**           | $(\frac{TP}{TP+FN})$                                       | Find actual positives        |
| **F1-Score**         | $(2\times\frac{Precision\times Recall}{Precision+Recall})$ | Balance Precision & Recall   |
| **Log Loss**         | $(-\frac{1}{n}\sum[y\log(p)+(1-y)\log(1-p)])$              | Model loss                   |



* **Logistic Regression** → Classification algorithm
* **Linear Equation** → Calculates score (**z**)
* **Sigmoid Function** → Converts score into probability (0 to 1)
* **Probability ≥ 0.5** → Class **1**
* **Probability < 0.5** → Class **0**
* **Log Loss** → Training loss function
* **Accuracy** → Overall performance
* **Precision** → Correct positive predictions
* **Recall** → Finds actual positives
* **F1-Score** → Best metric for imbalanced datasets


```python
import numpy as np
import pandas as pd
```

```python
df=pd.read_csv('/content/covid_toy - covid_toy.csv')
df.head(2)
```

```python
df['fever']=df['fever'].fillna(df['fever'].mean())
df.head(2)
```

```python
from sklearn.preprocessing import LabelEncoder
lb=LabelEncoder()
df['cough']=lb.fit_transform(df['cough'])
df['gender']=lb.fit_transform(df['gender'])
df['city']=lb.fit_transform(df['city'])
df['has_covid']=lb.fit_transform(df['has_covid'])
df.head(2)
```

```python
x=df.drop(columns=['has_covid'])
y=df['has_covid']
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
from sklearn.linear_model import LogisticRegression
lr=LogisticRegression()
lr.fit(X_train,y_train)
```

```python
y_pred=lr.predict(X_test)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(y_test,y_pred)
```
